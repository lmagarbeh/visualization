```python
## making random sample data
np.random.seed(613)
x=np.arange(50)
y=np.random.randint(0,100,50)
```

```python
### making a scatterplot
fig, ax=plt.subplots(figsize=(5,3)) ## 5 is the width and 3 is the height (size of the figure)
ax.scatter(x,y) ### creates two objects, the figure or the whole canvas, and the canvas is where you draw
```

```python
fig, ax=plt.subplots(figsize=(5,3)) 
ax.bar(x,y) 
```

```python
fig, ax=plt.subplots(figsize=(5,3)) 
ax.plot(x,y) 
```

```python
fig, ax=plt.subplots(figsize=(5,3)) 
ax.hist(y, edgecolor="black") ### we do not put x, because we need the frequency of the y value only
```

```python
fig, ax=plt.subplots(figsize=(5,3)) 
ax.plot(x,y) 
ax.set_title("Total growth over time")
ax.set_ylabel("Total growth")
ax.set_xlabel("Years since start")
fig.tight_layout()

```

```python
font1={'family': 'sans-serif', 'color': 'blue', 'size': 20} #### set up font colors for easier changes later on
font2={'family': 'monospace', 'color': 'green', 'size': 14}
```

```python
fig, ax=plt.subplots(figsize=(5,3)) 
ax.plot(x,y) 
ax.set_title("Total growth over time", fontdict=font1)
ax.set_ylabel("Total growth",fontdict=font2 )
ax.set_xlabel("Years since start", fontdict=font2)
fig.tight_layout()
```

```python
fig, ax=plt.subplots (figsize=(5,3))

ax.scatter(x,  ## changing from circle to star for the points
           y, 
           marker="*", 
           color="indigo")

fig.show() ### sometimes you need fig.show to show the figure
```

```python
fig, ax=plt.subplots (figsize=(5,3))

ax.plot(x,  ## changing from circle to star for the points
           y, 
           marker="*", 
           color="indigo", 
           linestyle="--",
           linewidth=2) ### specifying the line style (dashline) and line width 

fig.show() ### sometimes you need fig.show to show the figure
```

```python
fig, ax=plt.subplots (figsize=(5,3))

### adding color

ax.plot(x,  ## changing from circle to star for the points
           y, 
           marker="*",
           color="#7425b9",
            linestyle="--",
           linewidth=2)
fig.show()
```

```python
fig, ax=plt.subplots (figsize=(5,3))

### adding color

ax.plot(x,  ## changing from circle to star for the points
           y, 
           marker="*",
           markersize=12,
           color="#7425b9",
            linestyle="--",
           linewidth=2, 
           markeredgecolor= '#fa9359', ## edge of the star
           markerfacecolor= "#000000") ## color inside star
fig.show()

ax.grid(axis='y') ### adding grid lines

## for the scatter plot, we do not have these options 
```

```python
fig, ax=plt.subplots (figsize=(5,3))

### adding grid lines
ax.grid(axis='y', color="blue", linewidth=2, linestyle='-.') ### adding grid lines


ax.plot(x,  ## changing from circle to star for the points
           y, 
           marker="*",
           markersize=12,
           color="#7425b9",
            linestyle="--",
           linewidth=2, 
           markeredgecolor= '#fa9359', ## edge of the star
           markerfacecolor= "#000000") ## color inside star
fig.show()

```

```python
#Activity
# libraries
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
 
# Dataset:
a = pd.DataFrame({ 'group' : np.repeat('A',500), 'value': np.random.normal(10, 5, 500) })
b = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(13, 1.2, 500) })
c = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(18, 1.2, 500) })
d = pd.DataFrame({ 'group' : np.repeat('C',20), 'value': np.random.normal(25, 4, 20) })
e = pd.DataFrame({ 'group' : np.repeat('D',100), 'value': np.random.uniform(12, size=100) })
df = pd.concat([a,b,c,d,e])

# plot violin chart
sns.violinplot( x='group', y='value', data=df)

# add title
plt.title("Violin plot", loc="left")

# show the graph
plt.show()
```
