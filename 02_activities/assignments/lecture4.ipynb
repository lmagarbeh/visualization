```python
np.random.seed(613)
x1=np.arange(50)
y1=np.random.randint(0,75,50)
x2=np.array(['Luffy', 'Zoro', 'Nami','Usopp' ,"Sanji"]) ## more operations can be applied on the variables using an array
y2=np.array([110,180,240,99,220])
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2, nrows=1, figsize=(7,3)) ### two graphs into columns and separate graphs
ax1.scatter(x1,y1)
ax1.annotate("This is important", xy=(10,65), xytext=(30,65), arrowprops=dict(facecolor='black'))
ax2.bar(x2,y2)

## another way to formulate two graphs
#fig,ax=plt.subplots(ncols=2, nrows=1, figsize=(7,3))
#ax[0].scatter (x1,y1)
#ax[1].bar(x2,y2)
```

```python
### naming the plot as someaxes and then we can refer to each graph separately
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4)) ### mosaic defines where the plots will be located. ax3 is always on the right, ax1 is in the top left and ax2 is in the bottom left

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
#Modify the figure layout by setting labels to x-axis
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4))

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18)

### the lable for the first plot is not showing
```

```python
#adjust the layout of the subplots so the titles appear 

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4))

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
fig.tight_layout() #### here 
```

```python
#adjust the layout of the subplots so the titles appear -another way using constraints

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4),
layout = "constrained") ## here it will automatically adjust everything including the legent, better than the fig_tight_layout

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python

# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))


# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")

ax.plot(x, y2,
color = "red")
```

```python
### adding error bars
y2_sd = np.std(y2) ### standard deviation of the y-values

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "blue")

ax.errorbar(x, #our x values
y2, #our y values
yerr = y2_sd,### the standard deviation for the error bar
color='red',
fmt = "none") 
## fmt how data 
## fmt none (draws nothing), - or o-, o --how to define our datamarkers. 

### if use the error bar function alone, it will plot the datapoints for us. 
```

```python
### customizing the error bars
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
color = "red")

ax.errorbar(x,
y2,
yerr = y2_sd,
fmt = "none",
ecolor= "indigo",
elinewidth= 4,
capsize = 6,
capthick= 4 )
```
