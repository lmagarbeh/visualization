```python
### Visualization lecture 3 code 
## Leen Magarbeh

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
### making random numbers

np.random.seed(613)
x=np.arange(50)
y1=np.random.randint(0,100,50)
y2=np.random.randint(0,100,50)
```

```python
### two line plots in the same figure and axes

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

```python
### adding a legend for each linegraph

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower center')
fig.show()

## legend is displayed on the lower right in the axes area. 
### other options like center or upper right/left or upper center or lower right/left, upper left, center left/right, or lower centre
```

```python
## customizing the legend

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = True, #add frame around the legend
fontsize = 12, #change font size
ncol = 2, #specify number of columns
shadow = True )
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',  ### attaching the  upper left corner of the legend to the top right corner of the plot
bbox_to_anchor =(1, 1)) ###this is the position where to anchor the legend, 0,0 is the origin of the rectangle (bottom left of the rectangle, 1,1 is the top right)
fig.show()
```

```python
### changing to a scatter plot

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!") ### putting text on the graph (annotation) at position 10 on x-axis and 95 on y-axis
fig.show()
```

```python
#annotation
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right', prop={"family":"serif" ,"size": 12}) ## changing font for legend. 


ax.text(10, 95, "This value is important!",
ha='center', ## horizontal alignment of text (centre of the sentence is located at 10)
color = 'red', ##modify font colour
size = 20) ##modify font size

#fig.show() ## needed for python not jupyter notebook 
```

```python
# annotation
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData ) #--this is with reference to the data - this is the default function
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes ) ## relatively is to the whole axes (half way across the axes and 10% of the y-axis. )
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure ) ## this is with reference to the whole figure (Proportion of the whole figure)
```

```python
# annotating with arrows

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), ### this is the location of the dot and the also the text
arrowprops=dict(facecolor='black')) ### color of arrow
fig.show()
```

```python
#modifying the arrows

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!',
xy=(10, 95), xytext=(20, 94),
arrowprops = dict(arrowstyle = "<->",
color = "hotpink"))
### arrowstyle - wedge, fancy, <->, 
```

```python
### removing axes tick marks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
## adding tick marks at specific intervals
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
### rotating axes
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right') ### horizontal alignment of the x-labels (right or left of the ticks)
```

```python
### changing the font of the x-axis. 
font1 = {'family':'serif','color':'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
#ax.set_xlabel("Shiny New X Axis!", fontsize = 18, fontdict = font1)
plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```

```python
### changing the font of the x-axis. 
font1 = {'family':'serif','color':'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.set_xlabel("Shiny New X Axis!", fontsize = 18, fontdict = font1) ## similar function to code above
#plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```
